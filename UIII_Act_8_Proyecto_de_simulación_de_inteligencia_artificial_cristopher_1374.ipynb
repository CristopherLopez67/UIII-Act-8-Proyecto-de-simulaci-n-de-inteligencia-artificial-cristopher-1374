```python
print("======================================")
print("   ASISTENTE IA DE ENFERMERIA")
print("======================================")

continuar = "si"

while continuar == "si":

    print("\nHola, soy tu asistente inteligente.")
    print("Te ayudare a elegir una jeringa y revisar sintomas.\n")

    nombre = input("¿Cual es tu nombre?: ")

    print("\nMucho gusto,", nombre)

    print("\nMENU PRINCIPAL")
    print("1. Revisar temperatura")
    print("2. Elegir jeringa")
    print("3. Revisar nivel de dolor")

    opcion = input("\nSelecciona una opcion: ")

    # OPCION 1
    if opcion == "1":

        temperatura = float(input("\nIngresa tu temperatura corporal: "))

        if temperatura >= 39:
            print("\nLa IA detecta fiebre alta.")
            print("Se recomienda acudir al medico.")

        elif temperatura >= 37:
            print("\nLa IA detecta fiebre leve.")
            print("Descansa y toma agua.")

        else:
            print("\nTemperatura normal.")
            print("Tu salud parece estable.")

    # OPCION 2
    elif opcion == "2":

        print("\nTIPOS DE JERINGAS")
        print("1 ml  -> Bebes")
        print("3 ml  -> Vacunas")
        print("5 ml  -> Medicamentos")
        print("10 ml -> Uso especial")

        jeringa = input("\n¿Que jeringa necesitas?: ")

        if jeringa == "1":
            print("\nLa IA recomienda jeringa de 1 ml.")
            print("Ideal para bebes.")

        elif jeringa == "3":
            print("\nLa IA recomienda jeringa de 3 ml.")
            print("Perfecta para vacunas.")

        elif jeringa == "5":
            print("\nLa IA recomienda jeringa de 5 ml.")
            print("Util para medicamentos.")

        elif jeringa == "10":
            print("\nLa IA recomienda jeringa de 10 ml.")
            print("Uso especial en hospitales.")

        else:
            print("\nOpcion no valida.")

    # OPCION 3
    elif opcion == "3":

        dolor = int(input("\nCalifica tu dolor del 1 al 10: "))

        if dolor >= 8:
            print("\nDolor muy fuerte.")
            print("La IA recomienda atencion inmediata.")

        elif dolor >= 5:
            print("\nDolor moderado.")
            print("Se recomienda descansar.")

        elif dolor >= 1:
            print("\nDolor leve.")
            print("Mantente en observacion.")

        else:
            print("\nNo hay dolor registrado.")

    # OPCION INVALIDA
    else:
        print("\nOpcion incorrecta.")

    continuar = input("\n¿Deseas continuar? (si/no): ")

print("\n======================================")
print(" Gracias por usar el sistema de IA")
print("======================================")
```
